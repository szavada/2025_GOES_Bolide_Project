```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
bd= pd.read_csv('bolide_data.csv') #bolide data csv provided by Jeff
```

```python
idlist= [1620242630952501001, 1720210531323209001, 1620203230752511001, 1620202841623396001, 1620202570754267001, 1720202540835450001, 1620202321503424001, 1620201991101138001, 1620201941040483001, 1620200961203175001, 1720200410543113001, 1620193571246140001, 1620193491433260001, 1720193411333028001, 1620193221354005001]
indexlist= [] #list of ids listed on the pdfs sent by Jeff
bdinfolist= []
```

```python
for i in range (len(bd)):
    for e in range (len(idlist)):
        if bd['_id'][i] == idlist[e]: #locate where the bolides are within bd
            indexlist.append(i)
```

```python
for i in range (len(indexlist)):
    bdinfolist.append(bd.iloc[indexlist[i]]) # make a list for every relevent bolide 
```
