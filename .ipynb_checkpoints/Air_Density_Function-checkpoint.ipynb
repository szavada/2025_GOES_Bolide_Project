```python
import numpy as np
```

```python
def airrho(h, rho0=1.225, H=8.5): #calculates density based on rho naught, height, and scale height
                                    #8.5 km is the scale height for Earth and 1.225 kg/m^3 is the density of air at sea level
    return ((1/(np.e**(h/H)))*rho0)
```
