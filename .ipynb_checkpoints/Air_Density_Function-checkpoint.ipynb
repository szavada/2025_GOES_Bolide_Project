```python
import numpy as np
```

```python
def airrho(h, rho0=1.225, H=8500): #calculates density based on rho naught, height, and scale height
                                    #8.5 km is the scale height for Earth and 1.225 kg/m^3 is the density of air at sea level
    return ((1/(np.e**(h/H)))*rho0)
```

```python
airrho(500)
```

```python
def degtom(theta, r=6378137): #r in m
    rad= theta*0.0175
    return r*rad
```

```python
def degtom2 (lat1, lat2, long1, long2, r= 6371.009): #haversine equation?
    lat1rad= lat1*0.0175
    lat2rad= lat2*0.0175
    long1rad= long1*0.0175
    long2rad= long2*0.0175

    dlat = lat2rad-lat1rad
    dlong = long2rad-long1rad
    a = (np.sin(dlat/2))**2
    b = (np.sin(dlong/2))**2
    inside = a + np.cos(lat1) * np.cos(lat2) * b
    d= 2*(np.arcsin(np.sqrt(inside))/np.arcsin(np.sqrt(1-inside)))*r #fix this should be 157.2257
    return(a, b, inside, d)
```

```python
degtom2(1, 2, 1, 2)
```

```python
degtom(.118)
```

```python
def findtheta(lat, long, alt):
    x= np.abs(lat[-1]-lat[0])*0.0175
    y= np.abs(long[-1]-long[0])*0.0175
    z= np.abs(alt[-1]-alt[0])
    return np.arctan((np.sqrt(x**2+y**2))/z)
```
