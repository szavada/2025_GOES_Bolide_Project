```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
bolide1= pd.read_csv('Bolide_1.csv', header=9)
```

```python
bolide1
```

```python
0.63/len(bolide1['time (ms)'])
```

```python
time1=(np.zeros_like(bolide1['time (ms)'])) #have to make an array because the entries are strings
```

```python
catdet=[] #made this in anticipation of needing to make time entries for the subsequent bolides
          #I was wrong

def times(timearray, timestep, bolidedata):
    for i in range(len(timearray)):
        if i > 0:
            timearray[i]= timearray[i-1]+timestep
        if bolidedata['energy (joules)'][i] == np.max(bolidedata['energy (joules)']):
            catdet.append(i)
```

```python
times(time1, .003, bolide1)
```

```python
plt.plot(time1[109:], bolide1['energy (joules)'][109:])
```

```python
time1[114]-time1[110] #base of the curve minus peak
```
