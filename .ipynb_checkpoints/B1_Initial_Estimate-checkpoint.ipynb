```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from ast import literal_eval
```

```python
lat= literal_eval(b1['lat_stereo_g16'])
long= literal_eval(b1['lat_stereo_g16'])
alt= literal_eval(b1['alt_stereo_g16'])
```

```python
print (np.abs(lat[-1]-lat[1]), np.abs(long[-1]-long[1]), np.abs(alt[-1]-alt[1]))
```

```python
def findtheta(lat, long, alt):
    x= np.abs(lat[-1]-lat[1])*0.0175
    y= np.abs(long[-1]-long[1])*0.0175
    z= np.abs(alt[-1]-alt[1])*0.0175
    return np.arctan((np.sqrt(x**2+y**2))/z)

def degtom(theta, r=6378137): #r in m
    rad= theta*0.0175
    return r*rad

def airrho(h, H=8500): #calculates density based on rho naught, height, and scale height
                                    #8.5 km is the scale height for Earth and 1.225 kg/m^3 is the density of air at sea level
    return (1/(np.e**(h/H)))

H=8500 #meters
rho0= 1.225 #kg/m^3
rockyrho = 3500 #kg/m^3
metalrho = 7500 #kg/m^3 #constants for bolide density based on possible composition
icyrho = 750 #kg/m^3
```

```python
findtheta(lat, long, alt)
```

```python
np.cos(findtheta(lat, long, alt))
```

```python
r= ((3*H*rho0)/(4*rockyrho*np.cos(findtheta(lat, long, alt))))* ((airrho(alt[-1])-airrho(alt[1])))
```
