```python
bdf = BolideDataFrame(source='glm')
bdf
```

```python
highbdf= bdf[bdf.confidenceRating == "high"]
```

```python
latdelta=highbdf.latitudeDelta #isolate latitude and longitude change 
longdelta=highbdf.longitudeDelta
```

```python
print(np.max(latdelta), np.max(longdelta))
```

```python
delta= np.sqrt(latdelta**2+longdelta**2) #use the pythagorean theorem to compute the actual track length of the bolides
```

```python
np.max(delta)
```

```python
bolidelist=[] #list of the longest tracked bolides 
latdeltalist=[] #the corresponding change in latitude for the longest tracked bolides
longdeltalist=[] #corresponding change in longitude, this will be used later to find the bolide locations in highbdf
indexlist=[] #tracks the placement of the bolides selected in delta

for i in range(len(delta)):   
    if delta.iloc[i]>3: #arbitrary number that shows the bolides with the greatest delta
        bolidelist.append(delta.iloc[i]) #append all values to their corresponding lists 
        latdeltalist.append(latdelta.iloc[i])
        longdeltalist.append(longdelta.iloc[i])
        indexlist.append(i)

bolidelist
```

```python
len(latdeltalist)
```

```python
bolidenumbers=[] #will hold the locations in highbdf of the bolides we want
w=0 
for i in range(len(highbdf)):
    for e in range (len(latdeltalist)): #search latdeltalist and longdeltalist for the corresponding values...
        if latdelta.iloc[i] == latdeltalist[e] and longdelta.iloc[i] == longdeltalist[e] and i not in bolidenumbers: #... and compare those values to make sure that specific bolide is the one we want
             bolidenumbers.append(i) #save the index 
```
