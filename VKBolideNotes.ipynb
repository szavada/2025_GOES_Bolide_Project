### Corresponding Midpoint values --> SFD curve (also an initial Chisq)

```python
from scipy.interpolate import interp1d

xsfd=np.linspace(0.1,5,1000)
ysfd=powerlaw(np.asarray(xsfd),A*fudgefactor,-3.4533697696073387)

Xcorrs= pd.read_csv("corresponding xpoints.csv", delimiter=',')
xXcorrs = Xcorrs.iloc[:, 0]
yXcorrs = Xcorrs.iloc[:, 1]
#compare yXcorrs to ysfd (at xXcorrs) for chi squared

print(f"len(xsfd): {len(xsfd)}")
print(f"len(ysfd): {len(ysfd)}")


# Interpolate ysfd to the xXcorrs points
interp_func = interp1d(xsfd, ysfd, bounds_error=False, fill_value=0)
ysfd_interp = interp_func(xXcorrs)

# Avoid division by zero in chi-squared
mask = ysfd_interp > 0
chi_squared = np.sum(((yXcorrs[mask] - ysfd_interp[mask])**2) / ysfd_interp[mask])

print("Chi-squared:", chi_squared)
```

### Chi-Squared pt.2 (very very high)

```python
CHI2 = 0
for yf, yf_model, ye in zip(yflash, ysfd_interp, yerr): 
    step = (yf - yf_model)**2 / (ye + 1e-49)**2  
    CHI2 += step

print(f"Chi-squared: {CHI2}")
```

### Reduced Chi Squared and P-value (+ fully formatted tail-only fit)

```python
#import curvefit for tail fit
from scipy.optimize import curve_fit


plt.style.use('dark_background')  # Apply dark background globally

def powerlaw(x, a, b):
    return a * x**b

# Tail cutoff
tail_cutoff = 0.5e-12
tail_mask = xXcorrs > tail_cutoff

x_tail = xXcorrs[tail_mask]
y_tail = yXcorrs[tail_mask]

# Clean tail data
valid_tail_mask = (x_tail > 0) & (y_tail > 0)
x_fit = x_tail[valid_tail_mask]
y_fit = y_tail[valid_tail_mask]

# Power-law fit
popt, pcov = curve_fit(powerlaw, x_fit, y_fit, p0=[1, -3])
a_fit, b_fit = popt
print(f"Fitted power-law (tail only): a = {a_fit:.3e}, b = {b_fit:.3f}")

# Interpolation for chi-squared
interp_func = interp1d(xsfd, ysfd, bounds_error=False, fill_value=0)
ysfd_interp = interp_func(x_tail)

valid_chi_mask = ysfd_interp > 0
chi_squared_tail = np.sum(
    ((y_tail[valid_chi_mask] - ysfd_interp[valid_chi_mask])**2) / ysfd_interp[valid_chi_mask]
)
reduced_chi_squared = chi_squared_tail / np.sum(valid_chi_mask)

print(f"Tail-only chi-squared: {chi_squared_tail:.2f}")
print(f"Tail data points used: {np.sum(valid_chi_mask)}")
print(f"Reduced chi-squared: {reduced_chi_squared:.2f}")


fig, ax = plt.subplots(figsize=(10, 6), facecolor='black')
ax.set_facecolor('black')

# Tail data
ax.errorbar(
    x_tail, y_tail, yerr=np.sqrt(y_tail),
    fmt='x', color='#FF6487', label='Tail Data',
    elinewidth=1.0, capsize=2, markersize=6
)

# SFD fit
line_fit, = ax.plot(xsfd, ysfd, linestyle='-', linewidth=2,
                    color='#FFC0CB', label='Power-law Fit (Tail)', alpha=0.85)


ax.set_xlabel('Peak Brightness', color='#F5BA98FF', fontsize=12)
ax.set_ylabel('Frequency', color='#F5BA98FF', fontsize=12)
ax.tick_params(axis='x', colors='white', labelsize=11)
ax.tick_params(axis='y', colors='white', labelsize=11)
ax.yaxis.label.set_color('#F5BA98FF')
ax.xaxis.label.set_color('#F5BA98FF')
ax.set_yscale('log')


ax.grid(True, color='gray', linestyle='--', linewidth=0.4, alpha=0.5)
for spine in ax.spines.values():
    spine.set_color('white')


legend = ax.legend(
    loc='upper right',
    frameon=True,
    facecolor='black',
    edgecolor='white',
    framealpha=0.9,
    fontsize=11
)
legend.get_texts()[0].set_color('#FFC0CB')  # Tail Data
legend.get_texts()[1].set_color('#FF6487')  # SFD line

plt.title('Tail-Only Power-law Fit', color='#F5BA98FF', fontsize=20)
plt.tight_layout()
plt.show()

```

```python
#regular version of my tail fit 
from scipy.optimize import curve_fit

tail_cutoff = 0.5e-12  # Adjust(?) -- I wanted a value smaller than the tick size for the lower x axis
tail_mask = xXcorrs > tail_cutoff

x_tail = xXcorrs[tail_mask]
y_tail = yXcorrs[tail_mask]

#remove zeros to avoid log(0) or division errors -- thanks chat
valid_tail_mask = (x_tail > 0) & (y_tail > 0)
x_fit = x_tail[valid_tail_mask]
y_fit = y_tail[valid_tail_mask]

#fitting the tail 
popt, pcov = curve_fit(powerlaw, x_fit, y_fit, p0=[1, -3])
a_fit, b_fit = popt
print(f"Fitted power-law (tail only): a = {a_fit:.3e}, b = {b_fit:.3f}")


#interpolate model at tail x-points for chi-squ
interp_func = interp1d(xsfd, ysfd, bounds_error=False, fill_value=0)
ysfd_interp = interp_func(x_tail)

#tail chi sq.
valid_chi_mask = ysfd_interp > 0
chi_squared_tail = np.sum(
    ((y_tail[valid_chi_mask] - ysfd_interp[valid_chi_mask])**2) / ysfd_interp[valid_chi_mask]
)

print(f"Tail-only chi-squared: {chi_squared_tail:.2f}")
print(f"Tail data points used: {np.sum(valid_chi_mask)}")
print(f"Reduced chi-squared: {chi_squared_tail / np.sum(valid_chi_mask):.2f}")

plt.figure(figsize=(8, 6))
plt.plot(xsfd, ysfd, 'r-', label='Power-law Fit (Tail)')
plt.errorbar(x_tail, y_tail, yerr=np.sqrt(y_tail), fmt='x', color='#251714FF', label='Tail Data')
plt.yscale('log')
plt.xlabel('Peak Brightness')
plt.ylabel('Frequency')
plt.title('Tail-Only Power-law Fit')
plt.grid(True, which='both') #formatting to make it look good
plt.legend()
plt.tight_layout()
plt.show()

 ### Colors (again): #FF9898FF, #A91E45FF, #D9636CFF, #251714FF
```
